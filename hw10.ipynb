```python
def ad_report(impressions, clicks, target=3.0):
    if impressions == 0:
        return (None, False)
    ctr = clicks / impressions * 100
    passed = ctr >= target
    return (ctr, passed)


def main():
    student_id = input("請輸入完整學號：").strip().upper()
    tail = student_id[-3:]
    a = int(student_id[-3])
    b = int(student_id[-2])
    c = int(student_id[-1])
    s = a + b + c

    ads = [
        ("搜尋廣告", 1000 + 100 * a, 30 + 5 * b),
        ("社群廣告", 1200 + 100 * b, 24 + 4 * c),
        ("影音廣告", 0, 0),
    ]
    target = 2.5 + 0.1 * c

    print(f"學號：{student_id}")
    print(f"CTR門檻：{target:.2f}%")

    passed_count = 0
    for name, impressions, clicks in ads:
        ctr, passed = ad_report(impressions, clicks, target=target)
        if passed:
            passed_count += 1
        ctr_text = "無曝光資料" if ctr is None else f"{ctr:.2f}%"
        print(
            f"{name}：曝光={impressions}，點擊={clicks}，"
            f"CTR={ctr_text}，達標={passed}"
        )

    print(f"達標廣告數：{passed_count}")
    print(f"預設門檻測試：{ad_report(1000, 30)}")
    print(f"零點擊測試：{ad_report(1000, 0)}")


if __name__ == "__main__":
    main()
```

```python
def collect_codes(raw_codes, saved=None):
    result = [] if saved is None else list(saved)
    for item in raw_codes:
        cleaned = item.strip().upper()
        if cleaned and cleaned not in result:
            result.append(cleaned)
    return result


def main():
    student_id = input("請輸入完整學號：").strip().upper()
    tail = student_id[-3:]
    a = int(student_id[-3])
    b = int(student_id[-2])
    c = int(student_id[-1])
    s = a + b + c

    saved = ["BASE"]
    raw_a = [
        " welcome" + tail + " ",
        "vip" + tail,
        "",
        "WELCOME" + tail,
        "   ",
    ]
    raw_b = [
        " new" + tail + " ",
        " welcome" + tail + " ",
    ]

    activity_a = collect_codes(raw_a, saved=saved)
    activity_b = collect_codes(raw_b)
    empty_result = collect_codes([])

    print(f"學號：{student_id}")
    print(f"活動A：{activity_a}")
    print(f"活動B：{activity_b}")
    print(f"原始saved：{saved}")
    print(f"原始raw_a：{raw_a}")
    print(f"空資料：{empty_result}")


if __name__ == "__main__":
    main()
```

```python
def cumulative_spend(daily_spend, n):
    if n == 0:
        return 0
    return cumulative_spend(daily_spend, n - 1) + daily_spend[n - 1]


def main():
    student_id = input("請輸入完整學號：").strip().upper()
    tail = student_id[-3:]
    a = int(student_id[-3])
    b = int(student_id[-2])
    c = int(student_id[-1])
    s = a + b + c

    daily_spend = [
        100 + 10 * a,
        150 + 10 * b,
        0,
        200 + 10 * c,
        50 + 10 * s,
    ]
    n = 3 + c % 3

    print(f"學號：{student_id}")
    print(f"每日費用：{daily_spend}")
    print(f"指定n：{n}")
    print(f"指定前{n}天累計：{cumulative_spend(daily_spend, n)}元")

    for days in range(0, 6):
        print(f"前{days}天：{cumulative_spend(daily_spend, days)}元")

    print(f"空資料：{cumulative_spend([], 0)}元")


if __name__ == "__main__":
    main()
```

```python
def select_clicks(raw_clicks, min_clicks=10):
    trimmed = map(lambda s: s.strip(), raw_clicks)
    non_empty = filter(None, trimmed)
    numbers = list(map(int, non_empty))
    qualified = list(filter(lambda x: x >= min_clicks, numbers))
    return (numbers, qualified)


def main():
    student_id = input("請輸入完整學號：").strip().upper()
    tail = student_id[-3:]
    a = int(student_id[-3])
    b = int(student_id[-2])
    c = int(student_id[-1])
    s = a + b + c

    raw_clicks = [
        str(10 + a),
        "0",
        "",
        " " + str(5 + b) + " ",
        "   ",
        str(15 + c),
        str(5 + b),
    ]
    min_clicks = 5 + b

    numbers, qualified = select_clicks(raw_clicks, min_clicks=min_clicks)
    total_qualified = sum(qualified)

    print(f"學號：{student_id}")
    print(f"點擊門檻：{min_clicks}")
    print(f"清理結果：{numbers}")
    print(f"合格結果：{qualified}")
    print(f"合格筆數：{len(qualified)}")
    print(f"合格合計：{total_qualified}")

    ranked = sorted(qualified, reverse=True)
    for rank, clicks in enumerate(ranked, start=1):
        print(f"第{rank}名：{clicks}次")

    _, high_qualified = select_clicks(raw_clicks, min_clicks=100)
    print(f"門檻100合格結果：{high_qualified}")
    print(f"門檻100合格筆數：{len(high_qualified)}")
    print(f"門檻100合格合計：{sum(high_qualified)}")
    if not high_qualified:
        print("無符合門檻的素材")


if __name__ == "__main__":
    main()
```
